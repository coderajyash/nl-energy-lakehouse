```python

import os

dbutils.widgets.text("repo_root", "/Workspace/Users/yashrajsingh.nl@gmail.com/nl-energy-lakehouse")
repo_root = dbutils.widgets.get("repo_root")
gold_dir = f"{repo_root}/sql/gold"

# Alphabetical order builds dim_* before fact_*, which the facts need.
for name in sorted(os.listdir(gold_dir)):
    if not name.endswith(".sql"):
        continue
    table = name.removesuffix(".sql")
    with open(f"{gold_dir}/{name}") as f:
        query = f.read().strip().rstrip(";")
    spark.sql(f"CREATE OR REPLACE TABLE workspace.gold.{table} AS {query}")
    print(f"built gold.{table}: {spark.table(f'workspace.gold.{table}').count()} rows")
```
